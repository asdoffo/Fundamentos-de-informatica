Ejercicio 1. Registro de temperaturas

```python
temperaturas = [25.5, 28.0, 22.5, 30.0, 27.5]

print("Temperaturas:")

for temperatura in temperaturas:
    print(temperatura)

#  promedio
promedio = sum(temperaturas) / len(temperaturas)
print("Temperatura promedio:", promedio)

#  máxima y mínima
print("Temperatura máxima:", max(temperaturas))
print("Temperatura mínima:", min(temperaturas))

#  nueva temperatura
temperaturas.append(26.0)

print("Nuevas temperaturas:", temperaturas)
```

Ejercicio 2. Notas de alumnos

```python
notas = [8, 7, 9, 6, 10, 7, 8]

#  promedio
promedio = sum(notas) / len(notas)
print("Promedio:", promedio)

# imprime los mayores o iguales a 8
contador = 0

for nota in notas:
    if nota >= 8:
        contador = contador + 1

print("La cantidad de notas mayores o iguales a 8 es:", contador)

# compruieba si el alumno aprobó o no
if promedio >= 6:
    print("Alumno aprobado")
else:
    print("Alumno desaprobado")

# ordena las notas 
notas.sort()

print("Notas ordenadas:", notas)
```

Ejercicio 3. Materiales de construcción

```python
materiales = ["cemento", "arena", "piedra", "hierro"]

# imprime los materiales
for material in materiales:
    print(material)

#agregar material ladrillos
materiales.append("ladrillos")

# quita arena
materiales.remove("arena")

# comprueba si hay cemento en la lista
if "cemento" in materiales:
    print("Hay cemento en la lista")
else:
    print("No hay cemento en la lista")

# cantidad de materiales
print("Cantidad de materiales:", len(materiales))

print("Lista final:", materiales)
```

Ejercicio 4. Coordenadas de un punto

```python
punto = (15, 25)

# ºcoordenada X
print("Coordenada X:", punto[0])

# coordenada Y
print("Coordenada Y:", punto[1])

#  distancia al origen
distancia = (punto[0]**2 + punto[1]**2)**0.5

print("Distancia al origen:", distancia)

# d) imprime las coordenadas
print("Punto:", punto[0], punto[1])
```

Ejercicio 5. Valores de una función

```python
valores = (10, 20, 30, 40, 50, 60)

# primer y último elemento
print("Primer elemento:", valores[0])
print("Último elemento:", valores[-1])

#  tres primeros elementos
print("Primeros tres elementos:", valores[:3])

# calcula cuántas veces aparece el número 30
cantidad = valores.count(30)

# imprime
print("Cantidad de veces que aparece 30:", cantidad)

# cantidad de elementos
print("Cantidad de elementos:", len(valores))
```

Ejercicio 6. Información de un alumno

```python
alumno = { 
    "nombre": "Juan",
    "edad": 20,
    "carrera": "Ingeniería Civil",
    "promedio": 8.5
}

# nombre
print("Nombre:", alumno["nombre"])

#  edad y carrera
print("Edad:", alumno["edad"])
print("Carrera:", alumno["carrera"])

# cambie el promedio
alumno["promedio"] = 9.0

#se agrega el año
alumno["año"] = 2026

# claves y valores
for clave, valor in alumno.items():
    print(clave, ":", valor)
```

Ejercicio 7. Inventario de materiales

```python
inventario = {
    "cemento": 50,
    "hierro": 120,
    "ladrillos": 1000,
    "arena": 30
}

#  inventario completo
for material, cantidad in inventario.items():
    print(material, ":", cantidad)

#  cantidad de cemento
print("Bolsas de cemento:", inventario["cemento"])

# cambia la cantidad de cemento
inventario["cemento"] = 65

# agrega ítem piedra
inventario["piedra"] = 80

# quita ítem arena
del inventario["arena"]

# cantidad total de unidades
total = sum(inventario.values())

print("Total de unidades almacenadas:", total)

print("Inventario actualizado:", inventario)
```
