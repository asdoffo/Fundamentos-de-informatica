## Solución usando listas y bucles para ahorrar código.

```python
# Datos de la obra
obra = input("Ingrese el nombre de la obra: ")
superficie = float(input("Ingrese los m2 de vereda a realizar: "))

# Lista con los materiales
materiales = ["Cemento", "Arena", "Piedras"]

# Lista con los consumosw necesarios por m2, en kg
consumo = [30, 18, 22]

# Con un bucle for se recorren las listas de materiales y consumos, solicitando al usuario las cantidades disponibles y calculando si hay suficiente material para la obra.
for i in range(3):

    print("\nMaterial:", materiales[i])

    # Pide las cantidades disponibles
    bolsas = int(input("Ingrese la cantidad de bolsas de " + materiales[i] + " disponibles: "))
    kg_bolsa = float(input("Ingrese los kg por bolsa de " + materiales[i]+ ": "))

    # Calculos
    disponible = bolsas * kg_bolsa
    necesario = superficie * consumo[i]

    # diferencia
    diferencia = disponible - necesario

    puede_comenzar = True

    # Impimie los resultados
    print("Cantidad disponible:", disponible, "kg")
    print("Cantidad necesaria:", necesario, "kg")

    if diferencia >= 0:
        print("Hay suficiente material.")
    else:
        print("Falta material.")
        print("Se necesitan comprar:", abs(diferencia), "kg")
        puede_comenzar = False

print("\n----------------------------------")
print("RESUMEN")
print("----------------------------------")

print("Obra:", obra.upper())
print("Superficie:", superficie, "m2")

if puede_comenzar:
    print("\nLa obra puede comenzar.")
    print("Hay material suficiente.")
else:
    print("\nLa obra no puede comenzar.")
    print("No hay material suficiente.")
```
